## **2. 단순회귀**

### **(1) 데이터 전처리**

#### **1) 데이터분할1 : x, y 나누기**

```
target = 'Sales'
feature = ['Price']
x = data.loc[:, feature]
y = data.loc[:, target]
```

#### **2) 데이터분할2 : train : validation 나누기**

```
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size = .3,
                                                  random_state = 20)
```

### **(2) 모델링**

* 하나의 feature로 tareget 예측
    * Price --> Sales

```
from sklearn.linear_model import LinearRegression
```

#### **1) 모델선언**

```
model1 = LinearRegression()
```

#### **2) 모델링(학습)**

```
model1.fit(x_train, y_train)
```

* 회귀계수 확인하기

```
print('회귀계수 : ', model1.coef_)
print('절편 : ', model1.intercept_)
```

#### **3) 검증 : 예측**

```
pred1 = model1.predict(x_val)
```

#### **4) 검증 : 평가**

```
# 평균오차
mean_absolute_error(y_val, pred1)
```

## **3. 회귀 모델 평가하기**

### **(1) 오차의 비로 평가하기 : R2 Score**

```
# r2 score로 평가하고 결과를 해석해 봅시다.
r2_score(y_val, pred1)
```

### **(2) 오차의 양과 율로 평가하기**

```
# MSE
mean_squared_error( y_val   , pred1   )
```

```
# RMSE
root_mean_squared_error( y_val   , pred1)
```

```
# MAE
mean_absolute_error(y_val   , pred1)
```

```
# MAPE
mean_absolute_percentage_error(y_val   , pred1)
```

## **4. 다중회귀**

### **(1) 데이터 전처리**

#### **1) 데이터분할1 : x, y 나누기**

```
target = 'Sales'
x = data.drop(target, axis=1)
y = data.loc[:, target]
```

#### **2) 가변수화**
모든 범주형 변수에 대해서 가변수화를 수행합니다.

```
cat_cols = ['ShelveLoc', 'US','Urban']
```

```
x = pd.get_dummies(x, columns=cat_cols, drop_first=True)
x.head()
```

#### **3) 데이터분할2 : train : validation 나누기**

```
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size = .3, random_state = 20)
```

### **(2) 모델링1**
Price + Age

```
# 학습에 사용할 변수를 지정합니다.
features = ['Price','Age' ]
x_train1 = x_train[features]
x_val1 = x_val[features]
```

```
x_train1.head()
```

```
model1 = LinearRegression()
model1.fit(x_train1, y_train)
```

```
print(list(x_train1))
print(model1.coef_, model1.intercept_)
```

```
pred1 = model1.predict(x_val1)
```

```
print('RMSE :', root_mean_squared_error(y_val, pred1))
print('MAE  :', mean_absolute_error(y_val, pred1))
print('r2 :', r2_score(y_val, pred1))
```

### (3) 모델링2
전체 feature 모두 사용

```
model3 = LinearRegression()
model3.fit(x_train, y_train)
```

```
print(list(x_train))
print(model3.coef_, model3.intercept_)
```

```
pred3 = model3.predict(x_val)
```

```
print('RMSE :', root_mean_squared_error(y_val, pred3))
print('MAE  :', mean_absolute_error(y_val, pred3))
print('r2 :', r2_score(y_val, pred3))
```
