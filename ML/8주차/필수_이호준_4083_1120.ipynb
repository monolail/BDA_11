## **3.범주형 X --> Y**

* 모든 범주형 X에 대해서 Y와 비교하여 차트를 그리고 수치화 하시오.
>   ① 시각화 : 평균비교 barplot  
>   ② 수치화 : t-test, anova

* 데이터를 분석하면서, 추가 분석할 만한 사항이 있는지 살펴보고 시도해보세요.

### **(1) ShelveLoc-> Sales**

① 시각화 : 평균비교 barplot

```python
sns.barplot(data=data, x='ShelveLoc', y='Sales')
plt.show()
```

② 수치화 : t-test, anova

```python
print(spst.ttest_ind(data[data['ShelveLoc']=='Bad']['Sales'], data[data['ShelveLoc']=='Medium']['Sales']))

temp = data.loc[data['Sales'].notnull()]
P_1 = temp.loc[temp.ShelveLoc == 'Bad', 'Sales']
P_2 = temp.loc[temp.ShelveLoc == 'Good', 'Sales']
P_3 = temp.loc[temp.ShelveLoc == 'Medium', 'Sales']

print(spst.f_oneway(P_1, P_2, P_3))
```

③ 시각화와 수치화 결과로부터 파악한 내용을 적어 봅시다.

```python
# ShelveLoc는 Sales와 관계가 없다라고 할 수 없으며, 따라서 귀무가설을 기각한다. : 두 개의 범주의 관계는 있다.
```

### **(2) Urban -> Sales**

① 시각화 : 평균비교 barplot

```python
sns.barplot(data=data, x='Urban', y='Sales')
plt.show()
```

② 수치화 : t-test, anova

```python
print(spst.ttest_ind(data[data['Urban']=='Yes']['Sales'], data[data['Urban']=='No']['Sales']))

temp = data.loc[data['Sales'].notnull()]
P_1 = temp.loc[temp.Urban == 'Yes', 'Sales']
P_2 = temp.loc[temp.Urban == 'No', 'Sales']


print(spst.f_oneway(P_1, P_2))
```

③ 시각화와 수치화 결과로부터 파악한 내용을 적어 봅시다.

```python
# Urban과 Sales는 pvalue가 0.05보다 크므로, 둘 사이의 관계가 없다, 따라서 귀무가설을 기각할 수 없으며 자유도는 398이다. : 두 개의 범주의 관계는 없다.
```

### **(3) US-> Sales**

① 시각화 : 평균비교 barplot

```python
sns.barplot(data=data, x='US', y='Sales')
plt.show()
```

② 수치화 : t-test, anova

③ 시각화와 수치화 결과로부터 파악한 내용을 적어 봅시다.
